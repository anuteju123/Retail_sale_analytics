```python
def check_out(df,column):
  q1 = df[column].quantile(0.25)
  q3 = df[column].quantile(0.75)
  IQR = q3-q1
  ll = q1-1.5*IQR
  ul = q3+1.,5*IQR
  outliers =df(df[column]<ll)|(df[column]>ul)
  return outliers

```
