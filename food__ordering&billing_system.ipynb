project _ FOOD ORDERING & BILLING SYSTEM
features:1.menu display 2.multiple items ordering 3. cart 4. discount calculation
discount rule: 15% for bills > 1000; 10% for bills > 500;
otherwise no discount.
student extention:
add gst, delivery charge, coupon code and remove/ update cart items.

```python

print("\n--------FOOD ORDERING SYSTEM---------")

total = 0

while True:

    print("\n--FOOD MENU--")
    print("1. Burger       - Rs.200")
    print("2. Pizza        - Rs.400")
    print("3. Biryani      - Rs.300")
    print("4. French Fries - Rs.150")
    print("5. Coke         - Rs.100")
    print("6. Finish Order")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        quantity = int(input("Enter quantity: "))
        total = total + (200 * quantity)
        print("Burger added")

    elif choice == 2:
        quantity = int(input("Enter quantity: "))
        total = total + (400 * quantity)
        print("Pizza added")

    elif choice == 3:
        quantity = int(input("Enter quantity: "))
        total = total + (300 * quantity)
        print("Biryani added")

    elif choice == 4:
        quantity = int(input("Enter quantity: "))
        total = total + (150 * quantity)
        print("French Fries added")

    elif choice == 5:
        quantity = int(input("Enter quantity: "))
        total = total + (100 * quantity)
        print("Coke added")

    elif choice == 6:
        break

    else:
        print("Invalid choice")


# DISCOUNT

if total > 1000:
    discount = total * 15 / 100
    print("15% discount applied")

elif total > 500:
    discount = total * 10 / 100
    print("10% discount applied")

else:
    discount = 0
    print("No discount applied")


# AMOUNT AFTER DISCOUNT

amount_after_discount = total - discount


# GST

gst = amount_after_discount * 5 / 100


# FINAL AMOUNT

final_amount = amount_after_discount + gst


# PAYMENT

print("\nPAYMENT METHOD")
print("1. Cash")
print("2. UPI")
print("3. Card")

payment = int(input("Choose payment method: "))

if payment == 1:
    payment_method = "Cash"

elif payment == 2:
    payment_method = "UPI"

elif payment == 3:
    payment_method = "Card"

else:
    payment_method = "Invalid"


# FINAL BILL

print("\n---- BILL ---- ")
print("Total Amount       :", total)
print("Discount           :", discount)
print("Amount After Disc. :", amount_after_discount)
print("GST (5%)           :", gst)
print("Final Amount       :", final_amount)
print("Payment Method     :", payment_method)
print("\n-----THANK YOU!-----")
```
