```python
#7.tuple- tuple unpacking
student = ("rahul",22,85,90,78)
#wpc to 1.extract the student name, 2.extract the markes,3.calculate the average,4. print whether the student passed or failed
student = ("rahul",22,85,90,78)
name = student[0]
marks = student[2:]
averege = sum(marks) / len(marks)
if all(m >=40 for m in marks):
  result = "passed"
else:
  result = "failed"
print("student name:",name)
print("marks:",marks)
print("averege:",averege)
print("result",result)
```
